## 3. Kiểu danh sách – list – list()


```python
sv = [369, 'Nguyễn Văn A', 7.8, 92]
print(sv)

# Hoặc viết dạng cột:
sv = [
 369,
 'Nguyễn Văn A',
 7.8,
 92
]
```

```python
print(sv[2])
print(sv[-3])

sv[1] = "Nguyễn Văn B"
print(sv)
```

### Các hàm có sẵn

- `list(iterable)`
- `len(lst)`
- `sum(lst)`
- `max(lst)`, `min(lst)`
- `sorted(lst, reverse=True)`
- `any(lst)`: Trả về True nếu có ít nhất một phần tử True.
- `all(lst)`: Trả về True nếu tất cả phần tử đều True.


```python
diem = [7.3, 9.5, 0, 4.6, 9.1]
print(any(diem)) # True
print(all(diem)) # False
```

### Phương thức liên quan

- `list.append(x)`
- `list.extend(iterable)`
- `list.insert(i, x)`
- `list.pop()`
- `list.remove(x)`
- `list.clear()`
- `list.sort(reverse=True)`


```python
numbers = [5, 2, 8]

numbers.append(10)
print(numbers)

numbers.remove(2)
print(numbers)

print(len(numbers))
print(max(numbers))
print(min(numbers))
print(sum(numbers))
```

### Ứng dụng trong phân tích dữ liệu

- Biểu diễn dữ liệu có cấu trúc: dạng bảng, JSON
- Áp dụng nhiều trong `pandas.DataFrame`

## 7. Kiểu bool – bool()

- Chỉ có 2 giá trị: `True`, `False`
- Độ chân trị (Truthiness):
  - **True**: khác `0`, không rỗng
  - **False**: số `0`, rỗng (`""`, `[]`, `()`, `{}`, `set()`), `None`


```python
print(True)
print(False)

print(10 > 5)
print(3 == 5)
print(5 != 2)
```

### Ép kiểu sang bool


```python
print(bool(1))
print(bool(0))
print(bool("Python"))
print(bool(""))
```

## TOÁN TỬ

### 1. Toán tử số học

- `+`, `-`, `*`, `/`
- `%`: Chia lấy dư
- `//`: Lấy phần nguyên
- `**`: Lũy thừa

### 2. Toán tử gán

- `=`: Phép gán
- `+=`, `-=`, `*=`, `/=`, `%=`, `//=`, `**=`: Tính toán rồi gán lại


```python
x = 5
x += 1
print(x)
```

### 3. Toán tử so sánh

- `>`, `<`, `>=`, `<=`, `==`, `!=` (Kết quả trả về kiểu `bool`)


```python
print(10 > 5)   # True (10 lớn hơn 5)
print(3 == 5)   # False (3 bằng 5 là sai)
print(5 != 2)   # True (5 khác 2)
```

### 4. Toán tử logic

- `and`: Phép và
- `or`: Phép hoặc
- `not`: Phép phủ định
- Kết quả trả về kiểu `bool`


```python
print(True and False)  # False (Cần cả hai cùng True)
print(True or False)   # True (Chỉ cần một trong hai là True)
print(not True)        # False (Phủ định của True)
```

### 5. Toán tử thành viên (Membership operators)

- `in` và `not in`: Kiểm tra sự tồn tại trong chuỗi, danh sách, tuple, dict (chỉ kiểm tra key), tập hợp.


```python
ds = [1, 2, 3]
print(2 in ds)         # True (2 có trong danh sách)
print(5 not in ds)     # True (5 không có trong danh sách)
# Dict (Từ điển - Chỉ kiểm tra Key, không kiểm tra Value):
info = {'ten': 'Hoang', 'tuoi': 20}
print('ten' in info)   # True ('ten' là một Key của từ điển)
print('Hoang' in info) # False ('Hoang' là Value chứ không phải Key)
```

### 6. Độ ưu tiên cơ bản

Thứ tự ưu tiên từ cao xuống thấp:
$$SỐ\ HỌC > SO\ SÁNH > LOGIC > GÁN$$

_Chú ý: Dùng dấu ngoặc `(...)` để thể hiện độ ưu tiên nếu muốn._


```python
x = 5 + 3 * 2 > 10 and not 2 == 3  # 11 > 10 and not False -> True and True -> True
print(x)  # True
# Dùng ngoặc (...) để đổi độ ưu tiên
print(2 + 3 * 4)    # 14 (Mặc định nhân trước)
print((2 + 3) * 4)  # 20 (Ép phải cộng trước)
```
