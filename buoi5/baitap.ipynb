## 3. Xử lý giá trị bị khuyết

Trong bộ dữ liệu, giá trị bị khuyết được biểu diễn bằng dấu `?`.
Chuyển dấu `?` thành `NaN` để Pandas nhận diện được giá trị thiếu.


```python
df.replace(
    to_replace="?",
    value=np.nan,
    inplace=True
)

df.head()
```

### Kiểm tra số lượng giá trị khuyết

```python
df.isna().sum()
```

### 3.1. Xóa các dòng bị thiếu ở cột `price`

Sử dụng `dropna()` với `subset=["price"]` và `axis=0` để loại bỏ dòng.


```python
df.dropna(
    subset=["price"],
    axis=0,
    inplace=True
)

df.shape
```

### 3.2. Thay giá trị thiếu của `horsepower` bằng giá trị trung bình

Trước tiên chuyển cột `horsepower` sang kiểu `float`, sau đó tính mean.


```python
df["horsepower"] = df["horsepower"].astype("float")

mean_horsepower = df["horsepower"].mean()

mean_horsepower
```

```python
df["horsepower"] = df["horsepower"].replace(
    to_replace=np.nan,
    value=mean_horsepower
)

df["horsepower"].isna().sum()
```

### 3.3. Thay giá trị thiếu của `normalized-losses` bằng mode

Mode là giá trị xuất hiện nhiều nhất trong cột.


```python
df["normalized-losses"] = df["normalized-losses"].astype("float")

mode_normalized_losses = df["normalized-losses"].mode()[0]

mode_normalized_losses
```

```python
df["normalized-losses"] = df["normalized-losses"].replace(
    to_replace=np.nan,
    value=mode_normalized_losses
)

df["normalized-losses"].isna().sum()
```

### 3.4. Xóa các dòng vẫn còn giá trị bị khuyết

Sau khi xử lý các cột theo yêu cầu, loại bỏ các dòng vẫn còn `NaN`.


```python
df.dropna(
    axis=0,
    inplace=True
)

df.shape
```

```python
df.isna().sum()
```

## 4. Định dạng dữ liệu

Chuyển các cột đang có kiểu dữ liệu chưa phù hợp sang kiểu số bằng `astype()`.


```python
df["normalized-losses"] = df["normalized-losses"].astype("int")
df["bore"] = df["bore"].astype("float")
df["stroke"] = df["stroke"].astype("float")
df["horsepower"] = df["horsepower"].astype("int")
df["peak-rpm"] = df["peak-rpm"].astype("int")
df["price"] = df["price"].astype("float")

df.dtypes
```

## 5. Chuyển giá trị số thành giá trị phân loại

Chia cột `price` thành ba nhóm:

- `Thấp`
- `Vừa`
- `Cao`

Dùng `np.linspace()` tạo 4 mốc để có 3 khoảng.


```python
bins = np.linspace(
    min(df["price"]),
    max(df["price"]),
    4
)

bins
```

```python
group_names = ["Thấp", "Vừa", "Cao"]

df["price-bin"] = pd.cut(
    df["price"],
    bins,
    labels=group_names,
    include_lowest=True
)

df[["price", "price-bin"]].head(10)
```

```python
df["price-bin"].value_counts()
```

## 6. Chuyển dữ liệu phân loại thành dữ liệu số

### 6.1. Integer Encoding cho cột `make`

Mỗi hãng xe được ánh xạ thành một số nguyên.


```python
make_mapping = {
    "alfa-romero": 0,
    "audi": 1,
    "bmw": 2,
    "chevrolet": 3,
    "dodge": 4,
    "honda": 5,
    "isuzu": 6,
    "jaguar": 7,
    "mazda": 8,
    "mercedes-benz": 9,
    "mercury": 10,
    "mitsubishi": 11,
    "nissan": 12,
    "peugot": 13,
    "plymouth": 14,
    "porsche": 15,
    "renault": 16,
    "saab": 17,
    "subaru": 18,
    "toyota": 19,
    "volkswagen": 20,
    "volvo": 21
}

df["make"] = df["make"].map(make_mapping)

df[["make"]].head(10)
```

### 6.2. Ordinal Encoding cho cột `num-of-doors`

Ánh xạ số cửa theo thứ tự:

- `two` thành `2`
- `four` thành `4`


```python
door_mapping = {
    "two": 2,
    "four": 4
}

df["num-of-doors"] = df["num-of-doors"].map(door_mapping)

df["num-of-doors"].value_counts()
```

### 6.3. One-hot Encoding cho các cột phân loại còn lại

Dùng `pd.get_dummies()` để chuyển mỗi giá trị phân loại thành các cột có giá trị `0` hoặc `1`.

Cột `price-bin` được giữ lại để quan sát kết quả phân nhóm giá.


```python
categorical_columns = [
    "fuel-type",
    "aspiration",
    "body-style",
    "drive-wheels",
    "engine-location",
    "engine-type",
    "num-of-cylinders",
    "fuel-system"
]

df = pd.get_dummies(
    df,
    columns=categorical_columns,
    dtype=int
)

df.head()
```

```python
df.shape
```

## 7. Chuẩn hóa dữ liệu bằng phương pháp Z-score

Công thức:

\[
x_{new} = \frac{x_{old} - \mu}{\sigma}
\]

Chuẩn hóa các cột số ban đầu của bộ dữ liệu.
Không chuẩn hóa các cột vừa được mã hóa và các cột One-hot có giá trị `0`, `1`.


```python
df["symboling"] = (
    df["symboling"] - df["symboling"].mean()
) / df["symboling"].std()

df["normalized-losses"] = (
    df["normalized-losses"] - df["normalized-losses"].mean()
) / df["normalized-losses"].std()

df["wheel-base"] = (
    df["wheel-base"] - df["wheel-base"].mean()
) / df["wheel-base"].std()

df["length"] = (
    df["length"] - df["length"].mean()
) / df["length"].std()

df["width"] = (
    df["width"] - df["width"].mean()
) / df["width"].std()

df["height"] = (
    df["height"] - df["height"].mean()
) / df["height"].std()

df["curb-weight"] = (
    df["curb-weight"] - df["curb-weight"].mean()
) / df["curb-weight"].std()

df["engine-size"] = (
    df["engine-size"] - df["engine-size"].mean()
) / df["engine-size"].std()

df["bore"] = (
    df["bore"] - df["bore"].mean()
) / df["bore"].std()

df["stroke"] = (
    df["stroke"] - df["stroke"].mean()
) / df["stroke"].std()

df["compression-ratio"] = (
    df["compression-ratio"] - df["compression-ratio"].mean()
) / df["compression-ratio"].std()

df["horsepower"] = (
    df["horsepower"] - df["horsepower"].mean()
) / df["horsepower"].std()

df["peak-rpm"] = (
    df["peak-rpm"] - df["peak-rpm"].mean()
) / df["peak-rpm"].std()

df["city-mpg"] = (
    df["city-mpg"] - df["city-mpg"].mean()
) / df["city-mpg"].std()

df["highway-mpg"] = (
    df["highway-mpg"] - df["highway-mpg"].mean()
) / df["highway-mpg"].std()

df["price"] = (
    df["price"] - df["price"].mean()
) / df["price"].std()

df.head()
```
