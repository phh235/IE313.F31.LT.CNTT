# Định dạng dữ liệu (Ví dụ thay thế giá trị)

Thay thế các giá trị chưa định dạng ở cột `Tam_tru` (`KTX`, `Ky tuc xa`, `Túc xá`, `Ký túc`) thành `Ký túc xá`.

```python
import pandas as pd

df = pd.DataFrame({
    'Ren_luyen': [78, 87, 78, 98, 100, 99],
    'Tam_tru': ['KTX', 'Ky tuc xa', 'Túc xá', 'KTX', 'Ký túc xá', 'Ký túc'],
    'diem': [5.7, 8.6, 2.7, 8.5, 9.3, 7.5]
})

print("Dữ liệu ban đầu:")
print(df)

df_replaced = df.replace(to_replace={
    'Tam_tru': ['KTX', 'Ky tuc xa', 'Túc xá', 'Ký túc']
}, value='Ký túc xá')

print("\nDữ liệu sau khi định dạng (replace):")
print(df_replaced)
```
