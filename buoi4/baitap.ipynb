### Yêu cầu:

Sử dụng bộ dữ liệu thô ở chương 1 (file `dataset_02.csv`). Sau đó, tiến hành xử lý theo quy trình sau:

1. Cột `price` => dropna => dòng
2. Cột `horsepower` => replace => mean
3. Cột `normalized-losses` => replace => mode
4. Bước cuối cùng, dropna dòng cho toàn bộ dữ liệu
5. Quan sát giá trị thực tế, sau đó chuyển kiểu phù hợp cho các cột còn lại.
   _Ví dụ: price là object thì chuyển thành Int64 hoặc float, tương tự cho các cột khác._


```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("dataset_02.csv")

print(f"Kích thước ban đầu: {df.shape}")
print("\n5 dòng đầu:")
print(df.head())
```

### Kiểm tra dữ liệu khuyết

Trong file, dữ liệu bị thiếu được ghi là `?`, nên đổi thành `NaN` để xử lý dễ hơn.


```python
df.replace('?', np.nan, inplace=True)

# Kiểm tra số lượng giá trị khuyết thiếu ở mỗi cột
print("Số lượng giá trị khuyết thiếu ở các cột:")
print(df.isna().sum())
```

### Bước 1: Xử lý cột `price`

Xóa các dòng bị thiếu giá trị `price`.


```python
df.dropna(subset=['price'], inplace=True)
print(f"Kích thước sau khi loại bỏ các dòng thiếu 'price': {df.shape}")
```

### Bước 2: Xử lý cột `horsepower`

Các giá trị thiếu của `horsepower` được thay bằng giá trị trung bình.


```python
# Chuyển cột horsepower sang dạng số
df['horsepower'] = df['horsepower'].astype('float')

# Tính giá trị trung bình của cột horsepower
mean_horsepower = df['horsepower'].mean()
print(f"Giá trị trung bình của horsepower: {mean_horsepower:.2f}")

# Thay giá trị thiếu bằng giá trị trung bình
df['horsepower'] = df['horsepower'].replace(to_replace=np.nan, value=mean_horsepower)
```

### Bước 3: Xử lý cột `normalized-losses`

Điền các giá trị thiếu bằng giá trị xuất hiện nhiều nhất (mode).


```python
# Chuyển cột normalized-losses sang dạng số
df['normalized-losses'] = df['normalized-losses'].astype('float')

# Tìm giá trị yếu vị (mode) của cột normalized-losses
mode_normalized_losses = df['normalized-losses'].mode()[0]
print(f"Giá trị yếu vị (mode) của normalized-losses: {mode_normalized_losses}")

# Thay giá trị thiếu bằng mode
df['normalized-losses'] = df['normalized-losses'].replace(to_replace=np.nan, value=mode_normalized_losses)
```

### Bước 4: Xóa các dòng còn thiếu dữ liệu

Sau khi xử lý các cột trên, xóa những dòng vẫn còn `NaN`.


```python
df.dropna(inplace=True)
print(f"Kích thước sau khi loại bỏ toàn bộ NaN còn lại: {df.shape}")
```

### Bước 5: Chuyển kiểu dữ liệu

Kiểm tra và chuyển các cột số đang có kiểu `object` về kiểu số phù hợp.


```python
print("Kiểu dữ liệu hiện tại:")
print(df.dtypes)
```

Chuyển `bore`, `stroke` sang `float`; các cột `normalized-losses`, `horsepower`, `peak-rpm` và `price` sang `int`.


```python
# Chuyển bore và stroke sang kiểu float
df['bore'] = df['bore'].astype('float')
df['stroke'] = df['stroke'].astype('float')

# Chuyển đổi các cột số nguyên sang int64
df['normalized-losses'] = df['normalized-losses'].astype('int64')
df['horsepower'] = df['horsepower'].astype('int64')
df['peak-rpm'] = df['peak-rpm'].astype('int64')
df['price'] = df['price'].astype('int64')

# Kiểm tra lại kiểu dữ liệu của DataFrame sau chuyển đổi
print("Kiểu dữ liệu sau khi chuyển kiểu:")
print(df.dtypes)
```

### Kiểm tra lại dữ liệu

Kiểm tra số ô còn thiếu và kiểu dữ liệu sau khi làm sạch.


```python
print("Số lượng khuyết thiếu còn lại:")
print(df.isna().sum())

print("\nThông tin DataFrame sạch:")
df.info()
```

### Bước 6: Chuẩn hóa dữ liệu

Chuẩn hóa ba cột `length`, `width`, `height` theo hai cách trong slide: Min-Max và Z-score.


```python
df[['length', 'width', 'height']].head()
```

#### Cách 1: Min-Max

Đưa các giá trị về khoảng từ 0 đến 1.


```python
df_minmax = df.copy()

df_minmax['length'] = (df['length'] - df['length'].min()) / (df['length'].max() - df['length'].min())
df_minmax['width'] = (df['width'] - df['width'].min()) / (df['width'].max() - df['width'].min())
df_minmax['height'] = (df['height'] - df['height'].min()) / (df['height'].max() - df['height'].min())

df_minmax[['length', 'width', 'height']].head()
```

#### Min-Max bằng `MinMaxScaler`


```python
from sklearn.preprocessing import MinMaxScaler

mm = MinMaxScaler()
df_minmax_sklearn = df.copy()

df_minmax_sklearn['length'] = mm.fit_transform(df[['length']])
df_minmax_sklearn['width'] = mm.fit_transform(df[['width']])
df_minmax_sklearn['height'] = mm.fit_transform(df[['height']])

df_minmax_sklearn[['length', 'width', 'height']].head()
```

#### Cách 2: Z-score

Sau khi chuẩn hóa, trung bình của mỗi cột sẽ gần bằng 0 và độ lệch chuẩn gần bằng 1.


```python
df_zscore = df.copy()

df_zscore['length'] = (df['length'] - df['length'].mean()) / df['length'].std()
df_zscore['width'] = (df['width'] - df['width'].mean()) / df['width'].std()
df_zscore['height'] = (df['height'] - df['height'].mean()) / df['height'].std()

df_zscore[['length', 'width', 'height']].head()
```
